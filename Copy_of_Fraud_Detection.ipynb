```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
import os
import kagglehub

# Download latest version and define path
path = kagglehub.dataset_download("mlg-ulb/creditcardfraud")

file_path = os.path.join(path, 'creditcard.csv')
df = pd.read_csv(file_path)

df.head()
```

```python
df.duplicated().sum()

df = df.drop_duplicates()
```

```python
df['Hour'] = df['Time'] // 3600
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
df['Amount_scaled'] = scaler.fit_transform(df[['Amount']])
```

```python
df['txn_freq'] = df.groupby('Hour')['Amount'].transform('count')  #transaction frequency
```

```python
df = df.drop(['Time', 'Amount'], axis=1)
```

```python
df.to_csv("cleaned_data.csv", index=False)
```

```python
plt.figure(figsize=(5, 4))
sns.countplot(x='Class', data=df)

plt.title("Fraud vs Normal Transactions")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 5))
sns.histplot(df[df['Class']==0]['Amount_scaled'], bins=50, color='blue', label='Normal')
sns.histplot(df[df['Class']==1]['Amount_scaled'], bins=50, color='red', label='Fraud')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
X=df.drop(columns=['Class'])
y=df['Class']
```

```python
from imblearn.over_sampling import SMOTE
smote = SMOTE(sampling_strategy='minority')
X_res, y_res = smote.fit_resample(X, y)
```

```python
y_res.value_counts()
```

```python
sns.countplot(x=y_res)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
   X_res, y_res, test_size=0.20, random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier

logistic_model=LogisticRegression()
```

```python
logistic_model.fit(X_train,y_train)
```

```python
y_pred_log=logistic_model.predict(X_test)
```

```python
from sklearn.metrics import confusion_matrix, classification_report,accuracy_score

# Confusion Matrix
conf_matrix = confusion_matrix(y_test, y_pred_log)
plt.figure(figsize=(8, 6))
sns.heatmap(conf_matrix, annot=True, fmt='d', cmap='Blues',
            xticklabels=['Not Fraud', 'Fraud'],
            yticklabels=['Not Fraud', 'Fraud'])
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.title('Confusion Matrix for Logistic Regression')
plt.show()
```

```python
# Classification Report
print('Classification Report for Logistic Regression:')
print(classification_report(y_test, y_pred_log))
```

```python
from sklearn.metrics import roc_curve, auc, accuracy_score, classification_report

scores = {}
acc = accuracy_score(y_test, y_pred_log)
scores['Accuracy'] = acc

# Calculate ROC curve and AUC
fpr, tpr, thresholds = roc_curve(y_test, logistic_model.predict_proba(X_test)[:, 1])
auc_score = auc(fpr, tpr)

# Plot ROC curve
plt.figure(figsize=(8, 6))
plt.plot(fpr, tpr, color='darkorange', lw=2, label=f'ROC curve (area = {auc_score:.2f})')
plt.plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--')
plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('Receiver Operating Characteristic (ROC) Curve')
plt.legend(loc='lower right')
plt.show()
```

```python
from sklearn.metrics import classification_report
import pandas as pd

# Get classification report as a dictionary
report_dict = classification_report(y_test, y_pred_log, output_dict=True)

# Extract metrics for Class 0 and Class 1
metrics_class_1 = {
    'Precision': report_dict['1']['precision'],
    'Recall': report_dict['1']['recall'],
    'F1-Score': report_dict['1']['f1-score'],
}

# Add overall accuracy
metrics_class_1['Accuracy'] = scores['Accuracy']

# Create a DataFrame for plotting
metrics_df = pd.DataFrame({
    'Metric': ['Precision', 'Recall', 'F1-Score', 'Accuracy'],
    'Class 1': list(metrics_class_1.values())
})

# Plotting
fig, ax = plt.subplots(figsize=(6, 4))

metrics_df.set_index('Metric').plot(kind='bar', ax=ax,)
ax.set_title('Logistic Regression Model Performance Metrics')
ax.set_ylabel('Score')
ax.set_xlabel('Metric')
ax.set_ylim(0, 1.05)
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
